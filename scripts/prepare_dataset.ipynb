```python
from pathlib import Path
from tqdm import tqdm
import shutil

print("Loading data...")

SOURCE_DIR = Path("../data/Input")
TARGET_DIR = Path("../data/sorted")

TARGET_DIR.mkdir(parents=True, exist_ok=True)

classes = {
    "crazing": "Crazing",
    "inclusion": "Inclusion",
    "patches": "Patches",
    "pitted": "Pitted_surface",
    "rolled": "Rolled_in_scale",
    "scratches": "Scratches"
}

for class_folder in classes.values():
    (TARGET_DIR / class_folder).mkdir(parents=True, exist_ok=True)

for file in tqdm(list(SOURCE_DIR.iterdir())):
    if not file.is_file():
        continue

    filename = file.name.lower()

    matched = False

    for keyword, class_folder in classes.items():
        if filename.startswith(keyword):
            destination = TARGET_DIR / class_folder / file.name
            shutil.copy2(file, destination)
            matched = True
            break

    if not matched:
        print("Not included:", file.name)

print("Data loaded.")
```
